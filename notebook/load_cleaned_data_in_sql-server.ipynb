```python
import pandas as pd 
import sqlite3
```

```python
conn = sqlite3.connect('ecommerce-sales.db')
```

```python
pd.read_sql("SELECT name FROM sqlite_master WHERE TYPE = 'table'",conn)
```

## **--------------Getting sales_info Tables Datas--------------------**

```python
complete_sales_info = pd.read_sql("SELECT * FROM sale_info",conn)
complete_sales_info
```

## **--------------Storing sales_info Tables Datas In  sql-server--------------------**

```python
import urllib
import logging
from sqlalchemy import create_engine


# ---------------------------------------------------------
# Logging Configuration
# ---------------------------------------------------------

logging.basicConfig(
    level=logging.INFO,
    filename = 'D:\project_pandas\project_01\log\load_cleaned_data_in_sql-server.logs',
    format="%(asctime)s | %(levelname)s | %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S",
    force = True
)

logger = logging.getLogger(__name__)


# ---------------------------------------------------------
# SQL Server Configuration
# ---------------------------------------------------------

SERVER = r"MADARA7998\SQLEXPRESS"
DATABASE = "ecommerce_db"
TABLE_NAME = "sales_info"
SCHEMA = "godl"


# ---------------------------------------------------------
# Create SQL Server Connection
# ---------------------------------------------------------

logger.info("Creating SQL Server connection...")

connection_string = (
    "DRIVER={ODBC Driver 17 for SQL Server};"
    f"SERVER={SERVER};"
    f"DATABASE={DATABASE};"
    "Trusted_Connection=yes;"
)

params = urllib.parse.quote_plus(connection_string)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={params}"
)

logger.info("SQLAlchemy engine created successfully.")


# ---------------------------------------------------------
# Upload DataFrame to SQL Server
# ---------------------------------------------------------

try:
    logger.info(
        "Starting data upload: %s.%s",
        SCHEMA,
        TABLE_NAME
    )

    logger.info(
        "Rows to upload: %s",
        len(complete_sales_info)
    )

    complete_sales_info.to_sql(
        name=TABLE_NAME,
        schema=SCHEMA,
        con=engine,
        if_exists="replace",
        index=False,
        chunksize=500
    )

    logger.info(
        "Data uploaded successfully to [%s].[%s]",
        SCHEMA,
        TABLE_NAME
    )
    print(f"Data uploaded successfully to scheme : {SCHEMA}  & table : {TABLE_NAME}" )

except Exception as e:
    logger.exception(
        "Error uploading data to SQL Server: %s",
        e
    )
    print("Error uploading data to SQL Server: %s",e)
```
